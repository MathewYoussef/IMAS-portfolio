# Phase 3–6 Consolidated Overview

This notebook aggregates diagnostics, congruence, multiblock, and baseline metrics so we can compare pipelines side-by-side.


```python
import json
from pathlib import Path
import pandas as pd

inputs = json.loads(Path('analysis/phase7/inputs.json').read_text())
phase3_corr = pd.read_json(inputs['phase3']['correlations']['path'])
phase3_reg = pd.read_csv(inputs['phase3']['regression_summary']['path'])
phase3_disp = pd.read_json(inputs['phase3']['dispersion']['path'])
phase4_cong = pd.read_csv(inputs['phase4']['congruence_summary']['path'])
phase4_dep = pd.read_csv(inputs['phase4']['dependence_metrics']['path'])
phase4_dcor = pd.read_csv(inputs['phase4']['dcor_summary']['path'])
phase5_multiblock = pd.read_csv(inputs['phase5']['multiblock_summary']['path'])
phase6_loocv = pd.read_csv(inputs['phase6']['baseline_loocv']['path'])

modality_summary = pd.read_csv('analysis/phase7/modality_summary.csv')
kind_modality_summary = pd.read_csv('analysis/phase7/kind_modality_summary.csv')

modality_summary.head(), kind_modality_summary.head()
```

```python
summary_cols = ['window', 'kind', 'modality', 'dcor_permutation_p', 'rv_permutation_p', 'baseline_rmse_loocv', 'baseline_mae_loocv', 'multiblock_joint_components']
comparison_table = kind_modality_summary[summary_cols].sort_values(['modality', 'kind'])
comparison_table
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 4))
comparison_table.pivot(index='kind', columns='modality', values='baseline_rmse_loocv').plot(kind='bar', ax=ax)
ax.set_ylabel('Baseline RMSE (LOOCV)')
ax.set_title('Phase 6 LOOCV by Kind/Modality')
ax.legend(title='Modality')
plt.tight_layout()
plt.show()
```

### Quick Observations

- Phase 3 permutation p-values remain high (see `phase3_corr.head()`), so linear correlations stay weak.
- Phase 4 dCor/RV values hover around 0.5–1.0 with p-values >0.09, indicating suggestive but non-significant alignment.
- Phase 5 stacked-PLS runs one joint component per modality; no warnings were raised.
- Phase 6 baseline LOOCV RMSE ranges from ~1.6 to ~3.7 (larger for Δ and 6 o’clock), reflecting overfit risk compared with in-sample RMSE.
